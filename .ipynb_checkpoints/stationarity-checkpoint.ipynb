```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def pdie(mu,n,N):
    return 1-(1-mu)**n
def pmut(mu,n,N):
    return 1-(1-mu)**((N-n))
```

```python
plt.plot(range(1,5), [pdie(0.01,i,4) for i in range(1,5)])
plt.plot(range(1,4), [pmut(0.01,i,4) for i in range(1,4)])
```

```python
def dp(p,r,d,mu,dt):
    #dp0 = r*p[0]*(1-p[0])-d*mu*p[0]-r*(1-p[0]-p[-1])*p[0]-r*mu*p[0]
    #dpi = [ r*p[i]*(1-p[i])+r*((i+1)-1)*mu*p[i-1]-d*(1-(1-mu)**(i+1))*p[i]-r*(1-p[i]-p[-1])*p[i]-r*np.sqrt(i+1)*mu*p[i] for i in range(1,len(p)-1)]
    #dpd = np.sum([d*(1-(1-mu)**(i+1))*p[i] for i in range(len(p)-1)])-r*p[-1]*(1-p[-1])
    n = len(p)-1
    dp1 = r*p[0]*(p[-1]-(d*pdie(mu,1,n)+pmut(mu,1,n)))
    dpi = [ r*p[i]*(p[-1]+(pmut(mu,i,n)*p[i-1]/p[i]-(d*pdie(mu,i+1,n)+pmut(mu,i+1,n)))) for i in range(1,n-1)]
    dpn = r*p[-2]*(p[-1]-d*pdie(mu,n,n))+r*pmut(mu,n-1,n)*p[-3]
    dpd = np.sum([r*d*pdie(mu,i+1,n)*p[i] for i in range(n)])-r*p[-1]
    
    dpis = np.concat([[dp1],dpi,[dpn],[dpd]])
    phi = np.sum(dpis)
    d_tot = np.array([dpis[i]-p[i]*phi for i in range(len(p))])
    return d_tot*dt
```

```python
p_list, dpis = [], []
p_init = [1/8]*8
p = np.copy(p_init)
p_list.append(p)
for i in range(5000):
    dpi = dp(p,r=0.5,mu=0.03,d=4,dt=0.1)
    p = p+dpi
    p_list.append(p)
    dpis.append(sum(dpi))
plt.plot(p_list,label=[str(i+1) for i in range(len(p)-1)]+["dead"])
#plt.ylim([0,1])
plt.legend()
plt.grid()
plt.show()
```
